```python
# Solution in Google Colab
cart_prices = [250.50, 120.00, 450.75, 80.25]

# 1. Nayi item add karein
cart_prices.append(199.50)
cart_prices.append(377.40)
cart_prices.append(100.80)

# 2. Aik item remove karein
cart_prices.remove(120.00)
cart_prices.remove(250.50)

# 3. Total items aur Total Bill calculate karein
total_items = len(cart_prices)
total_bill = sum(cart_prices)

print("New Prices:", cart_prices)
print("Final Total Items:", total_items)
print("Final Total Bill (PKR):", total_bill)
```

```python
# Solution in Google Colab
marks = [78, 92, 65, 88, 81]

# 1. Highest aur Lowest marks
highest = max(marks)
lowest = min(marks)

# 2. Marks ko ascending order mein sort karein
marks.sort()

# 3. Total aur Percentage (Total 500 marks)
obtained_marks = sum(marks)
percentage = (obtained_marks / 500) * 100

print("Sorted Marks:", marks)
print("Highest Marks:", highest, "| Lowest Marks:", lowest)
print("Percentage:", round(percentage, 2), "%")
```
